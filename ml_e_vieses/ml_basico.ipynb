```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# DADOS MOCADOS
dados = {
    "idade":[25,45,50,23,60],
    "renda":[3000,5000,8000,11000,2500],
    "comprou":[0,1,0,1,0]
}

df = pd.DataFrame(dados)

# COLUNAS DE ENTRADA
X = df[["idade", "renda"]]

# COLUNA DE INTERESSE (TARGET)
Y = df["comprou"]

# TREINAMENTO
modelo = LogisticRegression()
modelo.fit(X, Y)

# PREVISÃO
df["previsao"] = modelo.predict(X)

print(df)
```
