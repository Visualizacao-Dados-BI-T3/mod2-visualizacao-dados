```python
import pandas as pd

# Dados mocados incluindo a coluna idade
dados_com_idade = {
    "id": [1, 2, 3, 4, 5],
    "email": [
        "joao.silva@email.com",
        "maria.souza@email.com",
        "carlos.oliveira@email.com",
        "ana.santos@email.com",
        "lucas.lima@email.com"
    ],
    "telefone": [
        "(11) 98765-4321",
        "(21) 97654-3210",
        "(31) 96543-2109",
        "(41) 95432-1098",
        "(51) 94321-0987"
    ],
    "idade": [28, 35, 42, 19, 50],
    "cidade": [
        "São Paulo",
        "Rio de Janeiro",
        "Belo Horizonte",
        "Curitiba",
        "Porto Alegre"
    ],
    "produto": [
        "Smartphone",
        "Notebook",
        "Fone de Ouvido",
        "Monitor",
        "Teclado Mecânico"
    ],
    "valor": [1500.00, 3500.00, 150.00, 800.00, 250.00]
}

# Criando o DataFrame atualizado
df_completo = pd.DataFrame(dados_com_idade)

# Exibindo a nova tabela
df_completo
```

Generalização

```python
idade_mascarada = df_completo.copy()

# Ajustando os limites (bins) para coincidir com a quantidade de labels
idade_mascarada["faixa_etaria"] = pd.cut(idade_mascarada["idade"],
                                         bins=[0, 20, 30, 40, 50, 60, 70],
                                         labels=["0-19", "20-29", "30-39", "40-49", "50-59", "60-69"],
                                         right=False)

idade_mascarada
```

Agregação

```python
resumo = (
    df_completo.groupby("cidade", as_index=False)
    .agg(pessoas=("id", "nunique"),
         valor_medio=("valor", "mean"),
         valor_total=("valor", "sum"))
)
display(resumo)
```

Perturbação

```python
import numpy as np

rng = np.random.default_rng(42)

perturbada = df_completo[["cidade", "valor"]].copy()
perturbada["valor"] = (
    perturbada["valor"] + rng.normal(0, 10, len(perturbada))
).round(2)
# Lado a lado somente para ensinar com dados fictícios.
comparacao = pd.DataFrame({
    "original": df_completo["valor"],
    "com_ruido": perturbada["valor"]
})
display(comparacao.head())
print("Média original:", round(df_completo["valor"].mean(), 2))
print("Média alterada:", round(perturbada["valor"].mean(), 2))
```

Pseudoanonimização

```python
import uuid
associacao_restrita = df_completo[["id"]].copy()
associacao_restrita["codigo"] = [uuid.uuid4().hex for _ in range(len(df_completo))]

pseudonimizada = (
    df_completo.merge(associacao_restrita, on="id", validate="one_to_one")
    [["codigo", "cidade", "produto", "valor"]].copy()
)
display(pseudonimizada.head())
print("Tabela de associação existe:", not associacao_restrita.empty)
# Não exibimos a associação. Ela ainda existe na memória deste notebook.
```

k-anonimato

```python
quase_identificadores = ["cidade", "faixa_etaria"]
contagens = (
    idade_mascarada.groupby(quase_identificadores, observed=True, dropna=False)
    .size().reset_index(name="pessoas_no_grupo")
)
k = 3
display(contagens)
display(contagens[contagens["pessoas_no_grupo"] < k])
print("Menor grupo nesse recorte:", contagens["pessoas_no_grupo"].min())
```

```python
saida_quase_identificadores = (
    idade_mascarada[quase_identificadores]
    .merge(contagens, on=quase_identificadores, validate="many_to_one")
)
saida_quase_identificadores = saida_quase_identificadores.loc[
    saida_quase_identificadores["pessoas_no_grupo"] >= k,
    quase_identificadores
].copy()
display(saida_quase_identificadores)
print("Registros retidos:", len(saida_quase_identificadores))
```

```python
import pandas as pd
from IPython.display import display

amostra = pd.DataFrame({
    "pessoa_id": [f"A{i:03}" for i in range(1, 25)],
    "nome": [f"Aluno fictício {i}" for i in range(1, 25)],
    "email": [f"aluno{i}@example.invalid" for i in range(1, 25)],
    "idade": [23, 27, 31, 35, 41, 45] * 4,
    "regiao": ["Centro"] * 18 + ["Periferia"] * 6,
    "modalidade": ["Online"] * 16 + ["Presencial"] * 2
                  + ["Online"] * 2 + ["Presencial"] * 4,
    "canal_resposta": ["Aplicativo"] * 16 + ["Balcão"] * 2
                     + ["Aplicativo"] * 2 + ["Balcão"] * 4,
    "informacao_saude": ["Relato fictício de saúde"] * 24,
    "matriculou": [True] * 24,
    "concluiu": [True] * 14 + [False] * 4
                + [True] * 2 + [False] * 4,
    "consentiu_ofertas": [True, False, True, False, True, False] * 4,
    "revogou_ofertas": [False] * 23 + [True]
})

display(amostra.head())
```

Resposta da atividade

C1. Crie base_minima com identificador interno, região, modalidade, matrícula e conclusão. O identificador ajuda a contar pessoas na análise interna, mas não precisa aparecer no resumo final.


```python
amostra.columns
```

```python
base_minima = amostra[["pessoa_id", "regiao", "modalidade", "matriculou", "concluiu"]].copy()
base_minima
```

C2. Agrupe por região e modalidade. Conte pessoas distintas e conclusões. Como existe uma pessoa por linha, a soma dos booleanos de conclusão conta concluintes.


```python
resumo = base_minima.groupby(["regiao", "modalidade"]).agg(
    pessoas=("pessoa_id", "nunique"),
    conclusoes=("concluiu", "sum")
)

resumo
```

C3. Calcule o percentual de conclusão. O denominador deve ser o número de pessoas matriculadas representadas no recorte; nesta amostra todas estão matriculadas.


```python
resumo["% conclusão"] = 100 * (resumo["conclusoes"] / resumo['pessoas'])

resumo
```

C4. Localize grupos com menos de três pessoas. Use três apenas como alerta didático, não como garantia de anonimização ou exigência da LGPD.


```python
grupos_menor_que_tres = resumo.loc[resumo['pessoas'] < 3]
grupos_menor_que_tres
```

```python
grupo_maior_que_tres = resumo.loc[resumo['pessoas'] >= 3]
grupo_maior_que_tres
```
